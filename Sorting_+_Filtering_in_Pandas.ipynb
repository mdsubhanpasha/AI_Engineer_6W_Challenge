```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. LOAD DATA
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

# 2. OVERVIEW
print("Total Employees:", df.shape[0])
print("Total Columns:", df.shape[1])

# 3. OVERALL ATTRITION
print("\nOverall Attrition:")
print(df['Attrition'].value_counts(normalize=True)*100)

# 4. DEPARTMENT WISE
print("\nDepartment Wise Attrition %:")
dept_attr = df.groupby('Department')['Attrition'].value_counts(normalize=True).unstack()*100
print(dept_attr)

# 5. JOBROLE WISE - NUVVU ICHINA OUTPUT
print("\nJobRole Wise Attrition Count:")
print(df.groupby('JobRole')['Attrition'].value_counts())

# 6. KEY INSIGHTS
print("\nKEY INSIGHTS:")
print("1. Sales Representative: 39.7% attrition - HIGHEST")
print("2. Laboratory Technician: 23.9% attrition")
print("3. Manager & Research Director: <5% attrition")
```
