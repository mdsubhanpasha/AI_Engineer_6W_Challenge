```python
import pandas as pd
data = {
    'Department': ['Sales','Sales','R&D','R&D','HR','HR'],
    'JobLevel': [1,2,1,3,2,4],
    'Gender': ['Male','Female','Male','Female','Male','Female'],
    'Attrition': ['Yes','No','No','Yes','Yes','No'],
    'OverTime': ['Yes','No','Yes','No','Yes','No'],
    'MonthlyIncome': [4000,6000,5000,9000,3500,12000]
}
df = pd.DataFrame(data)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Department': ['Sales','Sales','R&D','R&D','HR','HR','Sales','R&D'],
    'JobLevel': [1,2,1,3,2,4,3,5],
    'Gender': ['Male','Female','Male','Female','Male','Female','Male','Female'],
    'Attrition': ['Yes','No','No','Yes','Yes','No','No','No'],
    'OverTime': ['Yes','No','Yes','No','Yes','No','Yes','No'],
    'MonthlyIncome': [4000,6000,5000,9000,3500,12000,8000,15000]
})

print("Q1 Department Salary:\n", df.groupby('Department')['MonthlyIncome'].mean())
print("\nQ2 Gender Attrition:\n", df.groupby(['Gender', 'Attrition']).size())
print("\nQ3 JobLevel Salary:\n", df.groupby('JobLevel')['MonthlyIncome'].mean())
print("\nQ4 Overtime Salary:\n", df.groupby('OverTime')['MonthlyIncome'].mean())

sns.barplot(x=df.groupby('Department')['MonthlyIncome'].mean().index,
            y=df.groupby('Department')['MonthlyIncome'].mean().values)
plt.title('Average Salary by Department')
plt.show()
```
